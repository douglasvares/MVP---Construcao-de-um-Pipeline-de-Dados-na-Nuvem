**8. Análise de Dados:**

**8.1. Receita, lucro e margem por mês:**

```python
%sql
SELECT
    d.ano_mes,
    ROUND(SUM(f.net_sales), 2) AS receita_liquida,
    ROUND(SUM(f.profit), 2) AS lucro_total,
    ROUND(
        100 * SUM(f.profit) / NULLIF(SUM(f.net_sales), 0),
        2
    ) AS margem_lucro_pct
FROM workspace.ecommerce_gold.fato_item_pedido f
JOIN workspace.ecommerce_gold.dim_data d
    ON f.sk_data = d.sk_data
GROUP BY d.ano_mes
ORDER BY d.ano_mes;
```

Os resultados mensais entre 2021 e 2025 evidenciam sazonalidade nas vendas, com picos recorrentes de receita e lucro no quarto trimestre, principalmente nos meses de novembro e dezembro. Esse comportamento sugere maior volume comercial em períodos associados a campanhas promocionais e às compras de fim de ano.

Entretanto, os meses de maior receita não necessariamente geraram maiores margens de lucro. Em novembro e dezembro, observa-se redução da margem percentual em comparação com outros meses, diminuindo que o crescimento das vendas pode estar relacionado à aplicação de descontos mais agressivos, a uma combinação de produtos menos rentáveis ​​ou a custos proporcionalmente maiores. Assim, a empresa deve acompanhar receita e lucro conjuntamente com margem, evitando avaliar o desempenho comercial somente pelo lucro.

**8.2. Categorias mais rentáveis:**

```python
%sql
SELECT
    p.product_category,
    SUM(f.quantity) AS quantidade_vendida,
    ROUND(SUM(f.net_sales), 2) AS receita_liquida,
    ROUND(SUM(f.profit), 2) AS lucro_total,
    ROUND(
        100 * SUM(f.profit) / NULLIF(SUM(f.net_sales), 0),
        2
    ) AS margem_lucro_pct
FROM workspace.ecommerce_gold.fato_item_pedido f
JOIN workspace.ecommerce_gold.dim_produto p
    ON f.sk_produto = p.sk_produto
GROUP BY p.product_category
ORDER BY lucro_total DESC;
```

A categoria **Eletrônicos** apresentou o maior desempenho em receita líquida e lucro total, demonstrando ser a principal categoria para geração de valor financeiro no conjunto analisado. Esse resultado é consistente com o maior valor unitário típico de itens como TVs, consoles, tablets, relógios inteligentes e smartphones.

Por outro lado, categorias como **Mercearia , Livros e Mídia , Material de Escritório , Beleza e Cuidados Pessoais e Brinquedos e Jogos** tiveram as maiores margens de lucro percentual. Portanto, embora não liderem necessariamente em faturamento total, essas categorias possuem boa eficiência de rentabilidade sobre as vendas realizadas.

A interpretação prática é que a Eletrônica deve receber atenção para manutenção de estoque, disponibilidade e campanhas, por seu impacto absoluto em receita e lucro. Já as categorias de maior margem podem ser exploradas para ampliar a rentabilidade, por exemplo, com ações de venda cruzada, recomendações personalizadas e campanhas direcionadas.

**8.3. Produtos com maior receita:**

```python
%sql
SELECT
    p.product_name,
    p.product_category,
    p.product_subcategory,
    SUM(f.quantity) AS quantidade_vendida,
    ROUND(SUM(f.net_sales), 2) AS receita_liquida,
    ROUND(SUM(f.profit), 2) AS lucro_total
FROM workspace.ecommerce_gold.fato_item_pedido f
JOIN workspace.ecommerce_gold.dim_produto p
    ON f.sk_produto = p.sk_produto
GROUP BY
    p.product_name,
    p.product_category,
    p.product_subcategory
ORDER BY receita_liquida DESC
LIMIT 10;
```

Os produtos com maior receita pertencem integralmente à categoria **Eletrônicos** , com destaque para itens como consoles de videogame, TVs, tablets, smartwatches e smartphones. Os produtos mais bem posicionados apresentaram volumes próximos de 670 a 780 unidades vendidas e receitas individuais aproximadas entre US$ 850 mil e US$ 957 mil.

O resultado mostra que os produtos tecnológicos de maior valor agregado têm papel central no faturamento do comércio eletrônico. Como possuem impacto relevante também no lucro, é importante acompanhar a disponibilidade desses itens, as condições comerciais negociadas com fornecedores e o nível de desconto aplicado.

Além disso, a concentração dos produtos líderes em uma única categoria representa uma oportunidade e um ponto de atenção: há potencial para discutir estratégias comerciais em Eletrônicos, mas a empresa deve evitar dependência excessiva dessa categoria e buscar oportunidades de crescimento em outras linhas de produtos.

**8.4. Desempenho por segmento de cliente:**

```python
%sql
SELECT
    c.customer_segment,
    c.region,
    COUNT(DISTINCT f.order_id) AS total_pedidos,
    ROUND(SUM(f.net_sales), 2) AS receita_liquida,
    ROUND(SUM(f.profit), 2) AS lucro_total,
    ROUND(
        SUM(f.net_sales) / NULLIF(COUNT(DISTINCT f.order_id), 0),
        2
    ) AS ticket_medio
FROM workspace.ecommerce_gold.fato_item_pedido f
JOIN workspace.ecommerce_gold.dim_cliente c
    ON f.sk_cliente = c.sk_cliente
GROUP BY
    c.customer_segment,
    c.region
ORDER BY receita_liquida DESC;
```

O segmento **Consumer** apresentou o melhor desempenho geral em receita e quantidade de pedidos, confirmando sua importância como principal público do negócio. Entre as regiões, a região **Sul** apareceu de forma consistente entre os líderes em receita e volume de pedidos.

Os ingressos médios ficaram, em geral, entre aproximadamente US$ 1,3 mil e US$ 1,5 mil, o que indica um padrão de valor por pedido relativamente estável entre os agrupamentos individuais. Dessa forma, os maiores negócios do segmento Consumidor e da região Sul parecem estar mais associados à frequência ou ao volume de pedidos do que a diferenças muito expressivas no ticket médio.

Com base nessas ações de manutenção e relacionamento devem priorizar os clientes Consumidor, especialmente na região Sul. Ao mesmo tempo, outros segmentos e regiões podem ser trabalhados com campanhas específicas para aumentar a recorrência e o ticket médio.

**8.5. Impacto dos descontos:**

```python
%sql
SELECT
    CASE
        WHEN discount_percentage = 0 THEN 'Sem desconto'
        WHEN discount_percentage > 0 AND discount_percentage <= 0.10 THEN 'Até 10%'
        WHEN discount_percentage > 0.10 AND discount_percentage <= 0.30 THEN 'De 10% a 30%'
        ELSE 'Acima de 30%'
    END AS faixa_desconto,
    COUNT(*) AS total_itens,
    ROUND(SUM(net_sales), 2) AS receita_liquida,
    ROUND(SUM(profit), 2) AS lucro_total,
    ROUND(AVG(profit_margin_percentage), 2) AS margem_media_pct
FROM workspace.ecommerce_gold.fato_item_pedido
GROUP BY
    CASE
        WHEN discount_percentage = 0 THEN 'Sem desconto'
        WHEN discount_percentage > 0 AND discount_percentage <= 0.10 THEN 'Até 10%'
        WHEN discount_percentage > 0.10 AND discount_percentage <= 0.30 THEN 'De 10% a 30%'
        ELSE 'Acima de 30%'
    END
ORDER BY margem_media_pct DESC;
```

A análise por faixa de desconto indica uma relação inversa entre desconto e rentabilidade: à medida que o percentual de desconto aumenta, a margem média de lucro diminui. A faixa **Sem desconto** apresentou margem maior, enquanto a faixa **Acima de 30%** apresentou margem percentual menor, apesar de ainda gerar receita.

Esse resultado demonstra que os descontos podem ajudar a sustentar ou aumentar o volume de vendas, mas melhorar a eficiência financeira de cada item vendido. Portanto, descontos elevados não devem ser utilizados de forma indiscriminada, principalmente em produtos que já possuem margem reduzida.

A recomendação é estabelecer regras comerciais por categoria, produto e faixa mínima de margem. Promoções acima de 30% serão avaliadas de forma criteriosa, podendo ser mais acertadas para liquidação de estoque, recuperação de vendas ou aquisição de clientes, desde que seu impacto no lucro seja monitorado.

**8.6. Desempenho por canal de vendas e marketing:**

```python
%sql
SELECT
    c.sales_channel,
    c.marketing_channel,
    c.campaign_name,
    COUNT(DISTINCT f.order_id) AS total_pedidos,
    ROUND(SUM(f.net_sales), 2) AS receita_liquida,
    ROUND(SUM(f.profit), 2) AS lucro_total
FROM workspace.ecommerce_gold.fato_item_pedido f
JOIN workspace.ecommerce_gold.dim_comercial c
    ON f.sk_comercial = c.sk_comercial
GROUP BY
    c.sales_channel,
    c.marketing_channel,
    c.campaign_name
ORDER BY lucro_total DESC;
```

Os canais **Mobile App** e **Website** concentraram os melhores resultados de receita e lucro. Entre os canais de marketing, **Organic Search** e **Direct** se destacaram, destacando que uma parcela relevante do desempenho comercial é proveniente de tráfego orgânico e de acessos diretos à plataforma.

Também foi identificado que muitos registros possuem campaign_name nulo. Isso sugere que uma parte expressiva das vendas não está associada a uma campanha especificamente identificada, reforçando a relevância dos canais de aquisição mais amplos, como acesso direto, busca orgânica e navegação pelos canais próprios.

Na prática, os resultados reforçam a necessidade de manter investimentos na experiência dos canais digitais, especialmente no aplicativo e no site. Além disso, seria aprimorado o preenchimento e o rastreamento das campanhas, pois a ausência do nome da campanha limita as análises mais planejadas sobre o retorno de ações de marketing específicas.

**8.7. Entregas, devoluções e avaliações:**

```python
%sql
SELECT
    p.delivery_status,
    p.return_status,
    COUNT(DISTINCT f.order_id) AS total_pedidos,
    ROUND(AVG(p.delivery_days), 2) AS media_dias_entrega,
    ROUND(AVG(p.customer_rating), 2) AS avaliacao_media
FROM workspace.ecommerce_gold.fato_item_pedido f
JOIN workspace.ecommerce_gold.dim_pedido p
    ON f.sk_pedido = p.sk_pedido
GROUP BY
    p.delivery_status,
    p.return_status
ORDER BY total_pedidos DESC;
```

Os pedidos com status de entrega **On Time** representam a maior parcela dos registros analisados, diminuindo a predominância de entregas realizadas dentro do prazo esperado. Esse resultado é positivo para a operação logística e para a experiência do cliente.

Os pedidos classificados como **Atrasado**s ​​avaliaram médias inferiores às entregas realizadas no prazo, evidenciando que atrasos possuem efeito negativo na percepção do consumidor. Por sua vez, as entregas convencionais como **Early** avaliaram as melhores avaliações médias, o que indica que a antecipação da entrega pode contribuir para maior satisfação.

Os registros cancelados ou devoluções não possuem valores de dias de entrega e avaliação, o que é coerente com situações em que a etapa de entrega ou a avaliação do cliente podem não ter sido concluídas. Para análises futuras, recomenda-se investigar os motivos das devoluções e cancelamentos separadamente, identificando produtos, regiões, modalidades de envio ou condições comerciais mais relacionadas a esses eventos.

**8.8. Clientes com maior potencial de fidelização:**

```python
%sql
SELECT
    c.customer_id,
    c.customer_name,
    c.customer_segment,
    COUNT(DISTINCT f.order_id) AS quantidade_pedidos,
    ROUND(SUM(f.net_sales), 2) AS receita_total,
    ROUND(SUM(f.profit), 2) AS lucro_total,
    ROUND(
        SUM(f.net_sales) / NULLIF(COUNT(DISTINCT f.order_id), 0),
        2
    ) AS ticket_medio,
    MAX(p.customer_lifetime_value) AS customer_lifetime_value,
    MAX(p.loyalty_points_earned) AS pontos_acumulados,
    MAX(p.loyalty_points_redeemed) AS pontos_resgatados
FROM workspace.ecommerce_gold.fato_item_pedido f
JOIN workspace.ecommerce_gold.dim_cliente c
    ON f.sk_cliente = c.sk_cliente
JOIN workspace.ecommerce_gold.dim_pedido p
    ON f.sk_pedido = p.sk_pedido
GROUP BY
    c.customer_id,
    c.customer_name,
    c.customer_segment
ORDER BY receita_total DESC
LIMIT 20;
```

Os clientes destacados no ranking apresentam características consistentes com alto potencial de fidelização: recorrência de compra entre 8 e 16 pedidos, receita elevada acumulada, lucro gerado, ticket médio expressivo, alto valor de vida do cliente (customer life value ) e movimentação relevante de pontos de fidelidade acumulados e resgatados.

Esses clientes são estratégicos porque já demonstraram relacionamento recorrente e contribuição financeira relevante para o negócio. O uso de indicadores de frequência, receita, ticket médio, CLV e pontos de fidelidade permite priorizar clientes com maior propensão à retenção e à recompra.

Como ação prática, a empresa pode direcionar benefícios personalizados a esse público, como ofertas exclusivas, acesso antecipado a promoções, bonificações no programa de pontos e comunicações segmentadas. Também é conversar com clientes com alto valor de vida, mas baixa frequência recente, pois eles podem representar risco de perda e devem ser incluídos em ações de reativação.

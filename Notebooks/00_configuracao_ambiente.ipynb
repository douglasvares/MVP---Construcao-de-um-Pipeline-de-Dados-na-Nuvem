**Criando uma tabela de registro de carga**

```python
%sql
CREATE SCHEMA IF NOT EXISTS workspace.ecommerce_bronze;
```

**Criando uma tabela de controle:**

```python
%sql
CREATE TABLE IF NOT EXISTS workspace.ecommerce_bronze.log_aquisicao_dados (
    id_carga BIGINT GENERATED ALWAYS AS IDENTITY,
    nome_arquivo STRING NOT NULL,
    fonte STRING NOT NULL,
    url_fonte STRING NOT NULL,
    caminho_arquivo STRING NOT NULL,
    formato STRING NOT NULL,
    camada_destino STRING NOT NULL,
    data_carga TIMESTAMP NOT NULL,
    observacao STRING
)
USING DELTA;
```
